```python
X = df[["vehicle_weight", "model_year"]][df.origin=="Asia"].head(7).values
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
dot = XTX_inv.dot(X.T)
w = dot.dot(y)
w
```
